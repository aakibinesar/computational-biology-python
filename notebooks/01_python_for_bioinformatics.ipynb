## 1. Biological sequences as Python strings

DNA, RNA, and protein sequences can be represented as strings.

For example, a DNA sequence is a string over the alphabet `{A, C, G, T}`.

```python
dna = "ATGCGATACGCTTGA"

print("DNA sequence:", dna)
print("Length:", len(dna))
print("First base:", dna[0])
print("Last base:", dna[-1])
print("First three bases:", dna[:3])
print("Last three bases:", dna[-3:])
```

## 2. Nucleotide counting

This connects to Rosalind problem `DNA`.

Given a DNA string, we want to count how many times each nucleotide appears.

```python
dna = "ACGTACGTGACG"

a_count = dna.count("A")
c_count = dna.count("C")
g_count = dna.count("G")
t_count = dna.count("T")

print(a_count, c_count, g_count, t_count)
```

The same task can be written as a reusable function.

```python
def count_nucleotides(sequence: str) -> dict[str, int]:
    """Count A, C, G, and T in a DNA sequence."""
    return {
        "A": sequence.count("A"),
        "C": sequence.count("C"),
        "G": sequence.count("G"),
        "T": sequence.count("T"),
    }


count_nucleotides("ACGTACGTGACG")
```

## 3. Dictionaries for sequence counting

This connects to Rosalind problem `INI6`, where dictionaries are used for word counting.

In bioinformatics, dictionaries are useful for counting nucleotides, codons, k-mers, motifs, and amino acids.

```python
from collections import Counter

dna = "ACGTACGTGACG"
counts = Counter(dna)

print(counts)
print("A:", counts["A"])
print("C:", counts["C"])
print("G:", counts["G"])
print("T:", counts["T"])
```

## 4. DNA to RNA transcription

This connects to Rosalind problem `RNA`.

Transcription can be represented computationally by replacing every `T` in a DNA sequence with `U`.

```python
def transcribe_dna_to_rna(dna: str) -> str:
    """Transcribe a DNA sequence into RNA by replacing T with U."""
    return dna.replace("T", "U")


dna = "GATGGAACTTGACTACGTAAATT"
rna = transcribe_dna_to_rna(dna)

print("DNA:", dna)
print("RNA:", rna)
```

## 5. Reverse complement

This connects to Rosalind problem `REVC`.

The complement rules for DNA are:

- `A` pairs with `T`
- `T` pairs with `A`
- `C` pairs with `G`
- `G` pairs with `C`

The reverse complement is obtained by complementing the sequence and reversing it.

```python
def reverse_complement(dna: str) -> str:
    """Return the reverse complement of a DNA sequence."""
    complement = {
        "A": "T",
        "T": "A",
        "C": "G",
        "G": "C",
    }

    return "".join(complement[base] for base in reversed(dna))


dna = "AAAACCCGGT"
print(reverse_complement(dna))
```

## 6. GC content

This connects to Rosalind problem `GC`.

GC content is the percentage of nucleotides in a DNA sequence that are either `G` or `C`.

```python
def gc_content(dna: str) -> float:
    """Return the GC content percentage of a DNA sequence."""
    if not dna:
        return 0.0

    gc_count = dna.count("G") + dna.count("C")
    return (gc_count / len(dna)) * 100


sequence = "AGCTATAG"
print(gc_content(sequence))
```

## 7. Motif search

This connects to Rosalind problem `SUBS`.

A motif is a smaller sequence pattern that appears inside a larger biological sequence.

The function below returns **one-based positions**, matching Rosalind-style output.

```python
def find_motif_positions(sequence: str, motif: str) -> list[int]:
    """Return one-based positions where motif appears in sequence."""
    positions = []

    for i in range(len(sequence) - len(motif) + 1):
        if sequence[i:i + len(motif)] == motif:
            positions.append(i + 1)

    return positions


sequence = "GATATATGCATATACTT"
motif = "ATAT"

print(find_motif_positions(sequence, motif))
```

## 8. Simple FASTA parsing

Many bioinformatics problems use FASTA-style input.

A FASTA record usually has:

- a header line beginning with `>`;
- one or more sequence lines.

The parser below is intentionally simple and suitable for beginner-level examples.

```python
fasta_text = ">Rosalind_1\nAGCTATAG\n>Rosalind_2\nCGATCGGCCT\n>Rosalind_3\nTTATAGCT\n"


def parse_fasta(text: str) -> dict[str, str]:
    """Parse simple FASTA text into a dictionary of ID -> sequence."""
    records = {}
    current_id = None
    current_sequence = []

    for line in text.strip().splitlines():
        line = line.strip()

        if line.startswith(">"):
            if current_id is not None:
                records[current_id] = "".join(current_sequence)

            current_id = line[1:]
            current_sequence = []
        else:
            current_sequence.append(line)

    if current_id is not None:
        records[current_id] = "".join(current_sequence)

    return records


records = parse_fasta(fasta_text)
records
```

## 9. Finding the sequence with highest GC content

This combines simple FASTA parsing with GC-content calculation.

It connects Python file/string handling to a realistic bioinformatics workflow.

```python
def highest_gc_record(fasta_records: dict[str, str]) -> tuple[str, float]:
    """Return the record ID and GC content of the sequence with highest GC content."""
    best_id = None
    best_gc = -1.0

    for record_id, sequence in fasta_records.items():
        current_gc = gc_content(sequence)

        if current_gc > best_gc:
            best_id = record_id
            best_gc = current_gc

    return best_id, best_gc


record_id, value = highest_gc_record(records)

print(record_id)
print(value)
```
